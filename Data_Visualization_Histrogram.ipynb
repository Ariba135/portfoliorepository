```python
import pandas as pd
import matplotlib.pyplot as plt

# Load your existing dataset
df = pd.read_csv(csv_files[0])

# Keep the two required columns and remove missing values
data = df[
    ["daily_screen_time_hours", "cgpa"]
].dropna()

# Create a vibrant 2D histogram
plt.figure(figsize=(10, 7))

plt.hist2d(
    data["daily_screen_time_hours"],
    data["cgpa"],
    bins=[18, 14],
    cmap="turbo",
    cmin=1
)

plt.colorbar(label="Number of Students")

plt.title(
    "Daily Screen Time vs CGPA",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("Daily Screen Time (Hours)", fontsize=12)
plt.ylabel("CGPA", fontsize=12)

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from google.colab import files

# Load your dataset
# Replace this path with your actual CSV filename if needed
df = pd.read_csv(csv_files[0])

# Select the two columns and remove missing/non-numeric values
data = df[["daily_screen_time_hours", "cgpa"]].copy()

data["daily_screen_time_hours"] = pd.to_numeric(
    data["daily_screen_time_hours"], errors="coerce"
)
data["cgpa"] = pd.to_numeric(data["cgpa"], errors="coerce")
data = data.dropna()

# Check that usable data exists
print("Number of valid rows:", len(data))
print(data.head())

if len(data) == 0:
    raise ValueError("No valid data found. Check the column names and CSV file.")

# Create the histogram
fig, ax = plt.subplots(figsize=(10, 7))

hist = ax.hist2d(
    data["daily_screen_time_hours"],
    data["cgpa"],
    bins=[18, 14],
    cmap="turbo",
    cmin=1
)

fig.colorbar(hist[3], ax=ax, label="Number of Students")
ax.set_title("Daily Screen Time vs CGPA", fontsize=16, fontweight="bold")
ax.set_xlabel("Daily Screen Time (Hours)")
ax.set_ylabel("CGPA")
ax.grid(False)

fig.tight_layout()

# Save the completed figure
filename = "/content/Screen_Time_vs_CGPA_Vibrant_Histogram.png"
fig.savefig(filename, dpi=300, bbox_inches="tight", facecolor="white")

# Display the chart, then download it
plt.show()
files.download(filename)
```
